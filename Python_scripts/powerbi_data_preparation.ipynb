STEP 1: Import Required Libraries

```python
import pandas as pd
import numpy as np
```


STEP 2: Data Collection (Load CSV Files)


```python
customers = pd.read_csv("D:\\SQL\\EDA SQL Project\\csv files\\dim_customers.csv")
products = pd.read_csv("D:\\SQL\\EDA SQL Project\\csv files\\dim_products.csv")
sales = pd.read_csv("D:\\SQL\\EDA SQL Project\\csv files\\fact_sales.csv")
```

 STEP 3: Data Understanding

STEP 4: Standardize Column Names (BEST PRACTICE)


```python
customers.columns = customers.columns.str.strip().str.lower()
products.columns = products.columns.str.strip().str.lower()
sales.columns = sales.columns.str.strip().str.lower()
```

STEP 5: Data Cleaning

```python
# Remove duplicates
customers.drop_duplicates(inplace=True)
products.drop_duplicates(inplace=True)
sales.drop_duplicates(inplace=True)
```

```python

# Handle missing values
customers.fillna("unknown", inplace=True)
products.fillna("unknown", inplace=True)
sales.fillna(0, inplace=True)
```

STEP 6: Data Type Conversion

```python
# Convert date columns
customers["birthdate"] = pd.to_datetime(customers["birthdate"], errors="coerce")
customers["create_date"] = pd.to_datetime(customers["create_date"], errors="coerce")
```

```python
products["start_date"] = pd.to_datetime(products["start_date"], errors="coerce")
```

```python
sales["order_date"] = pd.to_datetime(sales["order_date"], errors="coerce")
sales["shipping_date"] = pd.to_datetime(sales["shipping_date"], errors="coerce")
sales["due_date"] = pd.to_datetime(sales["due_date"], errors="coerce")
```

```python
# Convert numeric columns
sales["quantity"] = sales["quantity"].astype(int)
sales["sales_amount"] = sales["sales_amount"].astype(float)
sales["price"] = sales["price"].astype(float)

```

 STEP 7: Data Filtering


```python
# Remove invalid sales records
sales = sales[sales["quantity"] > 0]
sales = sales[sales["sales_amount"] > 0]
```

STEP 8: Feature Engineering

```python
# Create full customer name
customers["customer_name"] = (
    customers["first_name"].str.strip().str.lower() + " " +
    customers["last_name"].str.strip().str.lower()
)
```

```python
# Extract Year & Month already exist, ensure correctness
sales["order_year"] = sales["order_date"].dt.year
sales["order_month"] = sales["order_date"].dt.month
```

STEP 9: Data Transformation

```python
# Standardize text columns
customers["country"] = customers["country"].str.strip().str.lower()
customers["gender"] = customers["gender"].str.strip().str.lower()
customers["marital_status"] = customers["marital_status"].str.strip().str.lower()
```

```python
products["product_name"] = products["product_name"].str.strip().str.lower()
products["category"] = products["category"].str.strip().str.lower()
products["subcategory"] = products["subcategory"].str.strip().str.lower()
products["product_line"] = products["product_line"].str.strip().str.lower()
```

STEP 10: Outlier Detection & Treatment

```python
upper_limit = sales["sales_amount"].quantile(0.99)
sales = sales[sales["sales_amount"] <= upper_limit]
```

STEP 11: Data Validation

```python
print("Customers shape:", customers.shape)
print("Products shape:", products.shape)
print("Sales shape:", sales.shape)

print(sales.isnull().sum())
```

STEP 12: Export Clean Data for Power BI

```python
customers.to_csv("clean_dim_customers.csv", index=False)
products.to_csv("clean_dim_products.csv", index=False)
sales.to_csv("clean_fact_sales.csv", index=False)
```
